# Deliver & Govern: Agent Delivery Lab

A runnable governance tabletop for an enterprise agent portfolio. All data is synthetic and local.

```python
from dataclasses import dataclass
from statistics import mean

@dataclass
class Risk:
    name: str
    likelihood: int
    impact: int
    owner: str
    mitigation: str

    @property
    def score(self):
        return self.likelihood * self.impact

def status(score):
    return 'RED' if score >= 16 else 'AMBER' if score >= 9 else 'GREEN'

print('Delivery governance lab initialized.')
```

## 1. Agent Maturity & Delivery Lifecycle

```python
maturity = {
    'Strategy and value case': 3,
    'Data and knowledge readiness': 2,
    'Evaluation and safety': 2,
    'Platform and operations': 3,
    'Governance and adoption': 2,
}
lifecycle = ['Discover', 'Design', 'Build', 'Validate', 'Pilot', 'Scale', 'Retire']
current_stage = 'Pilot'
average_maturity = mean(maturity.values())
next_focus = [area for area, score in maturity.items() if score < 3]
print(f'Current lifecycle stage: {current_stage}')
print(f'Average maturity: {average_maturity:.1f}/5')
print('Priority uplift areas:', ', '.join(next_focus))
print('Lifecycle:', ' -> '.join(lifecycle))
```

## 2. Risk in Agentic Systems

```python
risks = [
    Risk('Prompt injection through retrieved content', 4, 5, 'Security lead', 'Isolate instructions, validate tool inputs, and red-team retrieval.'),
    Risk('Incorrect customer guidance', 3, 4, 'Product owner', 'Ground answers, evaluate quality, and provide human escalation.'),
    Risk('Sensitive-data disclosure', 2, 5, 'Privacy lead', 'Minimize data, enforce access controls, and redact logs.'),
    Risk('Tool outage or latency', 3, 3, 'Operations lead', 'Use timeouts, fallbacks, and a manual workflow.'),
]
for risk in sorted(risks, key=lambda item: item.score, reverse=True):
    print(f'[{status(risk.score)}] {risk.name}: {risk.score}/25 | Owner: {risk.owner}')
    print(f'  Mitigation: {risk.mitigation}')
```

## 3. Responsible AI & Governance Deep Dive

```python
release_evidence = {
    'Documented purpose and boundaries': True,
    'Human escalation path tested': True,
    'Safety evaluation pass rate >= 95%': True,
    'Privacy review approved': True,
    'Bias and accessibility test complete': False,
    'Production monitoring configured': True,
}
missing = [control for control, complete in release_evidence.items() if not complete]
decision = 'GO' if not missing else 'NO-GO'
print(f'Release decision: {decision}')
print('Open controls:', ', '.join(missing) if missing else 'None')
print('Governance record: retain evaluation results, approvals, model version, and rollback owner.')
```

## 4. Operating Agents at Scale

```python
service_metrics = {
    'availability_pct': 99.92,
    'p95_latency_seconds': 2.7,
    'tool_success_pct': 98.4,
    'human_escalation_pct': 6.8,
    'safety_block_pct': 0.3,
}
thresholds = {
    'availability_pct': (99.9, 'min'),
    'p95_latency_seconds': (3.0, 'max'),
    'tool_success_pct': (98.0, 'min'),
    'human_escalation_pct': (10.0, 'max'),
}
for metric, (target, direction) in thresholds.items():
    value = service_metrics[metric]
    healthy = value >= target if direction == 'min' else value <= target
    print(f"{'OK' if healthy else 'ALERT'} | {metric}: {value} (target {direction} {target})")
print('On-call policy: investigate any safety block spike, unavailable dependency, or breached SLO.')
```

## 5. Rollout, RAID & Delivery Tabletop

```python
raid = [
    ('Risk', 'Knowledge source may contain stale policy', 'Product owner', 'Weekly content review', 'AMBER'),
    ('Assumption', 'Human support team can absorb escalations', 'Service manager', 'Validate staffing before pilot', 'AMBER'),
    ('Issue', 'Accessibility test is incomplete', 'UX lead', 'Complete before release gate', 'RED'),
    ('Dependency', 'Identity group provisioning', 'IAM lead', 'Confirm access by pilot start', 'GREEN'),
]
for category, item, owner, action, flag in raid:
    print(f'[{flag}] {category}: {item} | Owner: {owner} | Next: {action}')

incident = {'event': 'Agent gives a policy answer without a cited knowledge source', 'severity': 'SEV-2'}
tabletop_actions = ['Disable the affected answer path', 'Route users to a human queue', 'Preserve traces and evidence', 'Correct the source and add a regression test', 'Approve rollback or re-release']
print(f"\nTabletop: {incident['severity']} - {incident['event']}")
for step, action in enumerate(tabletop_actions, 1):
    print(f'{step}. {action}')
```

## 6. Portfolio Governance & Reuse

```python
portfolio = [
    {'agent': 'Customer support', 'domain': 'Service', 'tools': {'knowledge_search', 'order_lookup'}, 'maturity': 3},
    {'agent': 'Employee IT help', 'domain': 'Internal IT', 'tools': {'knowledge_search', 'ticket_lookup'}, 'maturity': 2},
    {'agent': 'Policy advisor', 'domain': 'Risk', 'tools': {'knowledge_search'}, 'maturity': 2},
]
tool_usage = {}
for agent in portfolio:
    for tool in agent['tools']:
        tool_usage.setdefault(tool, []).append(agent['agent'])
print('Reusable capability candidates:')
for tool, agents in tool_usage.items():
    if len(agents) > 1:
        print('- {}: used by {}'.format(tool, ', '.join(agents)))
print('\nPortfolio steering decision: standardize shared knowledge_search controls and prioritize maturity uplift for Internal IT and Policy advisor.')
```
